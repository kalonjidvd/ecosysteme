# ATELIER TECHNIQUE DE PROGRAMMATION
## Du procedural à l’orienté objet

### 1. Introduction

#### 1.1. Objectifs

Les développeurs software ont découvert qu'utiliser une approche modulaire, dans
une conception et une implémentation orientée objet peut rendre les groupes de
développements software plus productifs que ne le permettait les techniques
précédentes. Les problèmes qui sont résolus dans une approche orientée objet sont
souvent plus faciles à analyser, à concevoir, à programmer, à maintenir et à faire
évoluer.

#### 1.2. Déscription générale

##### PARTIE 1 : Etude de cas d'un écosystème

###### MODULE 1 : Du procédural à l'orienté objet

A l'aide de la simulation d'un petit écosystème en Python notre objectif est d'apporter aux programmeurs du procédural une vue d'ensemble des mécanismes de la programmation et conception orientée objets et des solutions que cela a apporté aux limites de la programmation et conception procédurale.


### 2. Publique

Tout étudiant, programmeur ou concepteur du procédural interêssé par l’orienté objet.


### 3. Préréquis

Les variables-constantes, les tableaux, l’assignation, les opérateurs arithmétiques, les
opérateurs de comparaison, les branches, les itérations et les fonctions.


### 4. Benefices

A l’issu de cet échange technique vous allez comprendre :

* La stratégie de la conception orientée objet
  
* Le diagramme de classe UML2 : la classe, les membres de classes
    (propriétés et opérations), la visibilité des membres d’une classe,
    l’association des classes, les cardinalités des associations.

* Les types de base et les types dérivés

* Le polymorphisme

* Les classes abstraites et les classes concrètes

```python
# %load src/python/procedural/Jungle.py
"""
Created on Wed Aug 31 06:52:30 2016
Last modified on Fri Aug 27 20:31:03 2021

@author: Kalonji Tshidiangatshila
@email : kalonjidvd@gmail.com
"""

# Declaration des dictionnaires de donnees dans la portee globale de l application

# Le dictionnaire Eau
lePointDeau = {'type' : 'Eau', 'quantite' : 68}

# Le dicionnaire Plante
laPlante = {'type' : 'Plante', 'espece' : 'Manguier', 'taille' : 64}

# Le dictionnaire Proie
laProie = {'type' : 'Proie', 'espece' : 'Singe', 'energie' : 29}

# Le dictionnaire Predateur
lePredateur = {'type' : 'Predateur', 'espece' : 'Lion', 'energie' : 37}


# Declaration et implementation des fonctions de l application

def diminuer(laRessource, decroit):
    ''' Diminue la quantite de la plante ou de l eau selon la valeur passee
    comme argument pour le parametre decroit.
    @param laRessource
    @param decroit
     '''
    if(laRessource['type'] == 'Eau'):
        if(laRessource['quantite'] > decroit):
            laRessource['quantite'] -= decroit

    elif(laRessource['type'] == 'Plante'):
        if(laRessource['taille'] > decroit):
            laRessource['taille'] -= decroit


def boire(laFaune):
    ''' Permet au predateur et a la proie de s abreuver au point d eau.
    Elle decremente la quantite de l eau pour incrementer l energie de l animal
    qui boit.
    '''
    if(laFaune['type'] == 'Proie'):
        while laFaune['energie'] <= 50:
            if(lePointDeau['quantite'] > 19):
                diminuer(lePointDeau, 5)
                laProie['energie'] += 5
    elif(laFaune['type'] == 'Predateur'):
        while laFaune['energie'] <= 50:
            if(lePointDeau['quantite'] > 19):
                diminuer(lePointDeau, 3)
                lePredateur['energie'] += 3


def manger(laFaune, laRessource):
    ''' Permet au predateur de manger la proie et a la proie de manger la plante.
    Entraine une decrementation de la ressource (proie y compris) pour incrementer
    l'energie de l'animal qui mange.
    '''
    if(laFaune['type'] == 'Proie' and laRessource['type'] == 'Plante'):
        while laFaune['energie'] <= 70:
            if(laRessource['taille'] > 29):
                diminuer(laRessource, 1)
                laFaune['energie'] += 1
    elif(laFaune['type'] == 'Predateur' and laRessource['type'] == 'Proie'):
        if(laRessource['energie'] != 0):
            while laFaune['energie'] < 99:
                laFaune['energie'] += 1
                --laRessource['energie']
            laRessource['energie'] = 0


def evoluer():
    ''' Permet a la Jungle de faire evoluer les ressources : la plante pousse,
    tandis que l'eau s'evapore.
    '''
    lePointDeau['quantite'] -= 4
    laPlante['taille'] += 4

# Point d'entree de l'application.
def main():

    #Affichage des valeurs des variables de l'application
    print("\nVALEUR DE DEPART DES VARIABLES DE L'APPLICATION\n")

    print("la proie : espece = %s" % laProie['espece'] + ", energie = %d" % laProie['energie'] + "%")
    print("le predateur : espece = %s" % lePredateur['espece'] + ", energie = %d" % lePredateur['energie'] + "%")
    print("le point d'eau : quantite = %d" % lePointDeau['quantite'] + "%")
    print("la plante : taille = %d" % laPlante['taille'] + "%")

    #Demarage de l'application
    print("\nDEMARRAGE DE L'APPLICATION...")

    print("\nla proie boit de l'eau...\n")
    boire(laProie)

    print("La quantite de l'eau est : %d" % lePointDeau['quantite'] + "%")
    print("L'énergie de la proie est : %d" % laProie['energie'] + "%")

    print("\nle prédateur boit de l'eau...\n")
    boire(lePredateur)

    print("La quantite de l'eau est : %d" % lePointDeau['quantite'] + "%")
    print("L'énergie du prédateur est : %d" % lePredateur['energie'] + "%")

    print("\nla proie mange la plante...\n")
    manger(laProie, laPlante)

    print("La taille de la plante est : %d" % laPlante['taille'] + "%")
    print("L'énergie de la proie est : %d" % laProie['energie'] + "%")

    print("\nle prédateur mange la proie...\n")
    manger(lePredateur, laProie)

    print("L'énergie de la proie est : %d" % laProie['energie'] + "%")
    print("L'énergie du prédateur est : %d" % lePredateur['energie'] + "%")

    print("\nLes ressources evoluent...\n")
    evoluer()

    print("La quantite de l'eau est : %d" % lePointDeau['quantite'] + "%")
    print("La taille de la plante est %d" % laPlante['taille'] + "%\n")


if __name__ == '__main__':
    main()
```

```python
# %load src/python/primer_oop/Predateur.py
#!/usr/bin/python

"""
Created on Wed Aug 31 11:36:25 2016
Last modified on Wed Sept 01 21:13:03 2021

@author: david KALONJi
@email : kalonjidvd@gmail.com
"""

class Predateur:

    __espece = None # Doit etre initialisee lors de l appel du constructeur
    __energie = 37 # initialisation de l'energie dans la definition de la classe


    def __init__(self, espece):
        '''
            Constructeur de la classe Predateur.

            @param espece
        '''
        self.__espece = espece


    def getEspece(self, ):
        '''
            Accesseur en lecture pour l'attribut prive espece.
        '''
        return self.__espece


    def getEnergie(self, ):
        '''
            Accesseur en lecture pour l'attribut prive energie.
        '''
        return self.__energie


    def setEnergie(self, energie):
        '''
            Accesseur en ecriture pour l'attribut prive energie.

            @param energie
        '''
        self.__energie = energie


    def augmenterDEnergie(self, energie):
        '''
            Permet l'augmentation de l'energie du predateur de la valeur passee
            comme argument.

            @param energie
        '''
        self.__energie += energie


    # L'énergie du prédateur ne peut pas dépasser 100 par sa méthode boire
    def boire(self, lePointDeau):
        '''
            Permet a un predateur d'entrer en interaction avec un objet de type
            Eau qui lui est passe comme argument pour s'abreuver. Ce qui entraine
            une diminution de la quantite de l'eau et une augmentation de
            l'energie du predateur.

            @param lePointDeau
        '''
        while self.__energie <= 50:
            if(lePointDeau.getQuantite() > 19):
                lePointDeau.diminuer(3)
                self.augmenterDEnergie(3)


    def manger(self, laProie):
        '''
            Permet a un predateur d'entrer en interaction avec un objet de type
            Proie qui lui est passe comme arguement pour se nourrir. Ce qui entraine
            une diminution de l'energie de la proie et une augmentation de
            l'energie du predateur.

            @param laProie
        '''
        if(laProie.getEnergie() != 0):
            while self.getEnergie() <= 99:
                self.setEnergie(self.getEnergie()+1)
                laProie.setEnergie(laProie.getEnergie()-1)
            laProie.setEnergie(0)
```

```python
# %load src/python/primer_oop/Proie.py
#!/usr/bin/python

"""
Created on Wed Aug 31 11:22:43 2016
Last modified on Wed Sept 01 21:13:03 2021

@author: david KALONJi
@email : kalonjidvd@gmail.com
"""

class Proie:

    __espece = None # Doit etre initialisee lors de l appel du constructeur
    __energie = 29 # Initialisation de l'energie dans la definition de la classe


    def __init__(self, espece):
        '''
            Constructeur de la classe Proie.

            @param espece
        '''
        self.__espece = espece


    def getEspece(self, ):
        '''
            Accesseur en lecture pour l attribut prive espece.
        '''
        return self.__espece


    def getEnergie(self):
        '''
            Accesseur en lecture pour l attribut prive energie.
        '''
        return self.__energie


    def setEnergie(self, energie):
        '''
            Accesseur en ecriture pour l attribut prive energie.

            @param energie
        '''
        self.__energie = energie


    def augmenterDEnergie(self, energie):
        '''
            Permet l augmentation de l energie de la proie de la valeur passee
            comme argument.

            @param energie
        '''
        self.__energie += energie


    # L'energie de la proie ne peut pas depasser 100
    def boire(self, lePointDeau):
        '''
            Permet a une proie d entrer en interaction avec un objet de type Eau
            qui lui est passe comme argument pour s abreuver. Ce qui entraine une
            diminution de la quantite de l eau et une augmentation de l energie
            de la proie.

            @param lePointDeau
        '''
        while self.__energie <= 50:
            if(lePointDeau.getQuantite() > 19):
                lePointDeau.diminuer(5)
                self.augmenterDEnergie(5)


    # L energie de la proie ne peut pas depasser 100
    def manger(self, laPlante):
        '''
            Permet a une proie d'entrer en interaction avec un objet de type Plante
            qui lui est passe comme argument pour se nourrir. Ce qui entraine une
            diminution de la taille de la plante et une augmentation de l'energie
            de la proie.

            @param laPlante
        '''
        while self.__energie <= 71:
            if(laPlante.getTaille() > 29):
                laPlante.diminuer(1)
                self.augmenterDEnergie(1)
```

```python
# %load src/python/heritage_polymorphisme/Jungle.py
#!/usr/bin/python

"""
Created on Wed Aug 31 20:52:30 2016
Last modified on Mon Sept 13 19:40:56 2021

@author: david KALONJi
@email : kalonjidvd@gmail.com
"""

# Importations des membres des types derives concrets
from Plante import Plante
from Eau import Eau
from Proie import Proie
from Predateur import Predateur

# Fonction principale
def main():

    '''
        Instanciation des objets des types concrets par l'appel des
        constructeurs.
    '''
    lePointDeau = Eau(68)
    laPlante = Plante(64, "Manguier")
    laProie = Proie("Singe", 29)
    lePredateur = Predateur("Lion", 37)

    '''
        Creation d'une liste lesRessources a l'aide des curly braces pour contenir
        les objets Ressource qui seront sujets au polymorphisme.
    '''
    lesRessources = {}
    lesRessources[0] = lePointDeau
    lesRessources[1] = laPlante

    print("\n\nECOSYSTEM SIMULATION...\n")

    print("\nlaPlante : type = %s, quantite = %d"
        % (laPlante.getEspece(), laPlante.getQuantite()) + "%\n")

    print("lePointDeau : quantite = %d"
        % (lePointDeau.getQuantite())+ "%\n")

    print("laProie : espece = %s, energie = %d"
        % (laProie.getEspece(), laProie.getEnergie()) + "%\n")

    print("lePredateur : espece = %s, energie = %d"
        % (lePredateur.getEspece(), lePredateur.getEnergie()) + "%\n")

    print("\nla proie boit de l'eau...\n")
    laProie.boire(lePointDeau)

    print("la quantité de l'eau est : %d"
        % (lePointDeau.getQuantite()) + "%")
    print("L'énergie de la proie est : %d"
        % (laProie.getEnergie()) + "%\n")

    print("le prédateur boit de l'eau...\n")
    lePredateur.boire(lePointDeau)

    print("la quantité de l'eau est : %d"
        % lePointDeau.getQuantite() + "%")
    print("L'énergie du prédateur est : %d"
        % lePredateur.getEnergie() + "%\n")

    print("la proie mange la plante...\n")
    laProie.manger(laPlante)

    print("la quantite de la plante est : %d"
        % laPlante.getQuantite() + "%")
    print("L'énergie de la proie est : %d"
        % laProie.getEnergie() + "%\n")

    print("le prédateur mange la proie...\n")
    lePredateur.manger(laProie)

    print("l'énergie de la proie est : %d"
        % laProie.getEnergie() + "%")
    print("l'énergie du prédateur est : %d"
        % lePredateur.getEnergie() + "%\n")

    print("Les ressources evoluent...\n")
    # Mise en oeuvre du Polymorphisme
    i = 0
    while i < len(lesRessources):
        lesRessources[i].evoluer()
        i += 1

    print("la quantité de l'eau est : %d"
        % lePointDeau.getQuantite() + "%")
    print("la quantite de la plante est : %d"
        % laPlante.getQuantite() + "%")


if __name__ == '__main__':
    main()
```
