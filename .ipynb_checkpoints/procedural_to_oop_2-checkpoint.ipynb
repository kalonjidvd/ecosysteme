# ATELIER TECHNIQUE DE PROGRAMMATION
## Du procedural à l’orienté objet

### 1. Introduction

#### 1.1. Objectifs

Les développeurs software ont découvert qu'utiliser une approche modulaire, dans
une conception et une implémentation orientée objet peut rendre les groupes de
développements software plus productifs que ne le permettait les techniques
précédentes. Les problèmes qui sont résolus dans une approche orientée objet sont
souvent plus faciles à analyser, à concevoir, à programmer, à maintenir et à faire
évoluer.

#### 1.2. Déscription générale

##### PARTIE 1 : Etude de cas d'un écosystème

###### MODULE 2 : Quelques principes orienté objet plus avancés

Ce deuxième module nous fera avancer dans quelques pratiques qui font la force
du paradigme orienté objet. Quelques principes SOLID et design pattern.


### 2. Publique

Tout étudiant, programmeur ou concepteur du procédural interêssé par l’orienté objet.


### 3. Préréquis

Module 1 de l'atelier d'échange technique du procédural à l'orienté objet


### 4. Benefices

A l’issu de cet échange technique vous allez comprendre :

* Les classes abstraites et les classes concrètes

* Le multihéritage

* Quelques principes de la conception orientée objet : le principe de
    ségrégation des interfaces, le principe ouvert à l’extension-fermé à la
    modification, le principe de substitution de LISKOV.

* Les interfaces

* La composition des classes

* Les différences entre C++ et Java du point de vue de l’implémentation du
    paradigme orienté objet : l’abstraction des classes, les interfaces, le
    polymorphisme, l’allocation dynamique de la mémoire.
